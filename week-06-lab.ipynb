Usually you `pd.read_csv(...)`; here we build a small frame inline so the lab is
self-contained. Then look **before** you compute. Run the cell.

```python
import pandas as pd

df = pd.DataFrame({
    "city":   ["Cebu", "Davao", "Iloilo", "Baguio", "Tacloban"],
    "region": ["Visayas", "Mindanao", "Visayas", "Luzon", "Visayas"],
    "pop":    [964_000, 1_777_000, 457_000, 366_000, 251_000],
})
print(df.head())
print("\nshape:", df.shape)
print("\ndtypes:\n", df.dtypes)
```

## Part 1 &middot; Select columns

One name in single brackets gives a **Series**; a list of names in double brackets gives
a smaller **DataFrame**. Run the cell.

```python
one = df["pop"]              # a Series
two = df[["city", "pop"]]    # a DataFrame

print(type(one).__name__, "\n", one, "\n")
print(type(two).__name__, "\n", two)
```

## Part 2 &middot; Filter rows with a condition

A comparison on a column gives a `True/False` Series. Put it in the brackets and pandas
keeps the rows where it is `True`. Run the cell.

```python
big = df[df["pop"] > 500_000]
print("big cities:\n", big, "\n")

print("just the names:\n", big["city"].tolist())
```

**Answer here:**

## Part 3 &middot; Combine conditions

Use `&` for and, `|` for or &mdash; **not** the words `and`/`or` &mdash; and wrap each
condition in parentheses. Run the cell.

```python
vis = df["region"] == "Visayas"
big = df["pop"] > 300_000

print("Visayan AND big:\n", df[vis & big][["city", "pop"]], "\n")
print("count of matches:", (vis & big).sum())
```

2. Why must each condition be wrapped in parentheses? (Hint: `&` binds tighter than `>`.)
