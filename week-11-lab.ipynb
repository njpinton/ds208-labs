## Part 2 &middot; Isolate the failing case

To fix a bug, first make it happen reliably on the *smallest* input. Run the cell and
narrow it down. Run the cell.

```python
def to_int(x):
    return int(x)

data = ["10", "20", "thirty", "40"]
for item in data:
    try:
        print(item, "->", to_int(item))
    except ValueError as e:
        print(item, "-> FAILED:", e)
```

**Answer here:**

1. Exactly one item failed. Which one, and what is it about that value that `int()` cannot
   handle?
   &rarr; *your answer*

2. "Reproduce, then isolate." How did looping item-by-item help you pin the failure to a
   single value rather than blaming the whole list?
   &rarr; *your answer*
