## Part 1 &middot; A dictionary is a labelled drawer

A list finds things by position. A **dictionary** finds them by a **key** you choose
&mdash; a name, an id, a word. Run the cell.

```python
populations = {          # key: value
    "Cebu": 964_000,
    "Davao": 1_777_000,
    "Iloilo": 457_000,
}

print("Davao:", populations["Davao"])
populations["Baguio"] = 366_000     # add a new entry
print("cities on record:", list(populations.keys()))

for city, pop in populations.items():
    print(f"{city:8} {pop:>10,}")
```

## Part 2 &middot; Counting with a dictionary

The single most common data-science move: **tally how often each thing appears.** A
dictionary maps each thing to its running count. Run the cell.

```python
votes = ["yes", "no", "yes", "yes", "abstain", "no", "yes"]

tally = {}
for v in votes:
    if v not in tally:
        tally[v] = 0
    tally[v] = tally[v] + 1

print(tally)
```

## Part 3 &middot; Reading from a file

So far the data was typed into the cell. Real data lives in files. First we **write** a
small one, then **read** it back &mdash; the reading is the part you'll do forever.
Run the cell.

```python
# write a small text file (one reading per line)
text = "12\n7\n0\n19\n4\n"
with open("hours.txt", "w") as f:
    f.write(text)

# read it back, line by line
with open("hours.txt") as f:
    for line in f:
        print(repr(line))    # repr shows the hidden newline
```

**Answer here:**

1. `repr(line)` shows each value as something like `'12\n'`. What is the `\n`, and why
   would `int("12\n")` still work but `int("12 hours")` would not?
   &rarr; *your answer*

2. The `with open(...)` block closes the file for you when it ends. Why is letting Python

## Part 4 &middot; Put it together

Read the file, turn each line into a number, and total it &mdash; the whole point of
loading a file is to compute something from it. Run the cell.

```python
total = 0
count = 0
with open("hours.txt") as f:
    for line in f:
        line = line.strip()      # remove the trailing newline and spaces
        if line == "":
            continue             # skip blank lines
        total = total + int(line)
        count = count + 1

print("readings:", count)
print("total:", total)
print("average:", total / count)
```

1. Remove the `line = line.strip()` line and run again. Does it still work? Explain what

2. The file has a `0` in it. Is that a real reading or missing data dressed as zero? You
   can't tell from the file alone &mdash; so what should you do about it before averaging?
