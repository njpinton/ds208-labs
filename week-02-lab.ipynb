## Part 2 &middot; Doing something to every item

A loop saves you writing the same line twenty times. This one walks a list and keeps
a running total.

````python
hours = [3, 0, 5, 2, 8, 0, 4]

total = 0
days_worked = 0

for h in hours:
    total = total + h
    if h > 0:
        days_worked = days_worked + 1

print("Total hours:", total)
print("Days with any work:", days_worked)
print("Average across days worked:", total / days_worked)
````

### 3 &middot; The branches are in the wrong order *(fix)*
`grade(95)` should say `excellent`, but it says `passing`. Fix the order.

````python
def grade(score):
    if score >= 60:
        return "passing"
    elif score >= 90:
        return "excellent"
    elif score >= 75:
        return "good"
    return "needs work"

print(grade(95))   # FIX: should print "excellent"
````

<details>
<summary>Solution</summary>

An `if/elif` chain stops at the **first** true test. `95 >= 60` is true, so it never reaches `excellent`. Check the **highest** bound first:

```python
def grade(score):
    if score >= 90:
        return "excellent"
    elif score >= 75:
        return "good"
    elif score >= 60:
        return "passing"
    return "needs work"

print(grade(95))   # excellent
```

</details>

### 6 &middot; Guard against divide-by-zero *(fill)*
Add a guard so `percent(1, 0)` returns `None` instead of crashing.

````python
def percent(part, whole):
    ____                     # FILL: if whole is 0, return None
    return 100 * part / whole

print(percent(3, 4))   # 75.0
print(percent(1, 0))   # should be None, not a crash
````

<details>
<summary>Solution</summary>

Handle the bad input first, then compute:

