## Part 1 &middot; Making code choose

`if` runs a block only when something is true. `elif` and `else` cover the other cases.

## Part 2 &middot; Doing something to every item

A loop saves you writing the same line twenty times. This one walks a list and keeps
a running total.

```python
hours = [3, 0, 5, 2, 8, 0, 4]

total = 0
days_worked = 0

for h in hours:
    total = total + h
    if h > 0:
        days_worked = days_worked + 1

print("Total hours:", total)
print("Days with any work:", days_worked)
print("Average across days worked:", total / days_worked)
```

## Part 3 &middot; Naming a piece of work

A function is a name for something you do more than once. Write it once, use it
anywhere.

The cell below defines one and uses it three times. Run it, then add a fourth call
with your own numbers.

```python
def average(numbers):
    """Return the mean of a list of numbers, or 0 for an empty list."""
    if len(numbers) == 0:
        return 0
    return sum(numbers) / len(numbers)


print(average([3, 5, 4]))
print(average([10]))
print(average([]))          # the empty case, handled instead of crashing

# your own call here
```
