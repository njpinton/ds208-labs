## Part 0 &middot; figure, axes, labels

`plt.subplots()` hands back a **figure** (canvas) and an **axes** (the plot). You draw
and label on the axes. A chart without labels is a puzzle. Run the cell.

```python
import matplotlib.pyplot as plt

years = [2021, 2022, 2023, 2024, 2025]
sales = [12, 15, 14, 20, 24]

fig, ax = plt.subplots()
ax.plot(years, sales, marker="o")
ax.set_xlabel("Year")
ax.set_ylabel("Sales (millions)")
ax.set_title("Sales over time")
plt.show()
```

## Part 1 &middot; Bar for categories, histogram for spread

A **bar** compares separate categories; a **histogram** shows the spread of one
column. Run the cell.

```python
import matplotlib.pyplot as plt

regions = ["Luzon", "Visayas", "Mindanao"]
pops    = [57, 21, 26]
scores  = [72, 85, 90, 60, 75, 88, 91, 55, 79, 83]

fig, (a1, a2) = plt.subplots(1, 2, figsize=(9, 3))
a1.bar(regions, pops); a1.set_title("Population by region"); a1.set_ylabel("Millions")
a2.hist(scores, bins=5); a2.set_title("Score distribution"); a2.set_xlabel("Score")
plt.tight_layout(); plt.show()
```

**Answer here:**

1. The bars sit apart; the histogram bars touch. What does that visual difference say about
   what each chart measures?
   &rarr; *your answer*

2. Change `bins=5` to `bins=2` and rerun. How does the story the histogram tells change,
   and what does that warn you about bin choice?
   &rarr; *your answer*

## Part 2 &middot; seaborn from a DataFrame

seaborn reads a DataFrame and names columns for each axis, with sensible defaults. Run
the cell.

```python
import seaborn as sns, matplotlib.pyplot as plt, pandas as pd

df = pd.DataFrame({
    "area":   [315, 78, 43, 120, 200],
    "pop":    [964, 457, 1780, 366, 251],
    "region": ["Vis", "Vis", "Luz", "Luz", "Vis"],
})
sns.scatterplot(data=df, x="area", y="pop", hue="region")
plt.title("Population vs area"); plt.show()
```

**Answer here:**

1. `hue="region"` added colour with no extra code. What did it encode, and why can that
   reveal a pattern a plain scatter would hide?
   &rarr; *your answer*

2. A scatter needs *two* numbers per row. Why couldn't you use it to show a single
   column's distribution?

## Part 3 &middot; Keep it honest

Bar length encodes value, so bars must start at zero. Watch what a cropped axis does.
Run the cell.

```python
import matplotlib.pyplot as plt

labels = ["A", "B", "C"]
vals   = [98, 100, 102]

fig, (a1, a2) = plt.subplots(1, 2, figsize=(9, 3))
a1.bar(labels, vals);                 a1.set_title("y from 0 - honest")
a2.bar(labels, vals); a2.set_ylim(95, 103); a2.set_title("y cropped - misleading")
plt.tight_layout(); plt.show()
```

**Answer here:**

1. The two charts show the *same* numbers. Why does the right one make the differences look
   dramatic, and why is that misleading for a bar chart?
   &rarr; *your answer*

2. A line chart may crop its y-axis without lying, but a bar chart may not. What is the
   difference in what each chart uses to encode the value?
   &rarr; *your answer*
