## Part 1 &middot; Maths applies to every element

Operate on an array and NumPy does it to each element at once &mdash; no loop, no
`.append()`. Run the cell.

```python
import numpy as np
prices = np.array([100, 200, 300])

print("with 12% VAT:", prices * 1.12)
print("doubled:     ", prices + prices)
print("sqrt of [1,4,9]:", np.sqrt(np.array([1, 4, 9])))
```

## Part 2 &middot; Summaries and the axis

Reductions collapse an array to a number. On a 2-D array, `axis` chooses the direction
that disappears. Run the cell.

```python
import numpy as np
m = np.array([[1, 2, 3],
              [4, 5, 6]])

print("grand total:", m.sum())
print("per column: ", m.sum(axis=0))   # the row dimension collapses
print("per row:    ", m.sum(axis=1))   # the column dimension collapses
print("mean, std:  ", m.mean(), round(m.std(), 2))
```

**Answer here:**

1. `m.sum(axis=0)` gave three numbers; `m.sum(axis=1)` gave two. Explain, in terms of the
   2x3 shape, why the counts differ.
   &rarr; *your answer*

2. "The axis that disappears." Which axis do you pass to get one number *per column*?
   &rarr; *your answer*

## Part 3 &middot; The boolean mask

A comparison returns an array of `True/False`. Index with it and you keep only the
elements where it is `True`. This is filtering. Run the cell.

```python
import numpy as np
rain = np.array([0, 12, 4, 33, 0, 8])

print("mask > 0:  ", rain > 0)
print("wet days:  ", rain[rain > 0])
print("avg wet:   ", round(rain[rain > 0].mean(), 2))
print("dry days:  ", (rain == 0).sum())
```

**Answer here:**

1. `rain[rain > 0]` *keeps* values; `(rain == 0).sum()` *counts* them. Explain how summing
   a `True/False` array gives a count.
   &rarr; *your answer*

2. Write one line that returns the number of days with more than 10 mm of rain.
   &rarr; *your answer*
