### Slide 38 — A function is a name for a piece of work


Expected output:

```
4.0
0
```


````python
def average(numbers):
    """Mean of a list, or 0 if empty."""
    if len(numbers) == 0:
        return 0
    return sum(numbers) / len(numbers)

print(average([3, 5, 4]))
print(average([]))
````

### Slide 44 — A counter that always returns 1


Expected output:

```
1
```


````python
def count_passing(scores):
    for s in scores:
        passing = 0       # resets each time!
        if s >= 60:
            passing = passing + 1
    return passing

print(count_passing([80, 40, 90]))   # want 2
````

### Slide 48 — Handling the invalid case first

````python
def average(nums):
    if not nums:
        return 0          # empty case, handled
    return sum(nums) / len(nums)

print(average([]), average([2, 4]))
````

### Slide 51 — Validation before computation


Expected output:

```
75.0 None
```


````python
def percent(part, whole):
    if whole == 0:
        return None       # nothing to be a % of
    return 100 * part / whole

print(percent(3, 4), percent(3, 0))
````
