### Slide 36 — Filtering and finding the maximum by hand


Expected output:

```
['Ana', 'Carla']
Carla 91
```


````python
# who passed?
passing = []
for s in students:
    if s["score"] >= 80:
        passing.append(s["name"])
print(passing)

# who's top?
top = students[0]
for s in students:
    if s["score"] > top["score"]:
        top = s
print(top["name"], top["score"])
````

### Slide 41 — Reading a file and computing a summary


Expected output:

```
6.333333333333333
```


````python
total = count = 0
with open("hours.txt") as f:
    for line in f:
        line = line.strip()
        if not line:
            continue          # skip blanks
        total += int(line)
        count += 1

print(total / count)
````

````python
with open("summary.csv", "w", newline="") as f:
    w = csv.writer(f)
    w.writerow(["section", "mean_score"])
    w.writerow(["A", 82.0])
    w.writerow(["B", 78.0])

print(open("summary.csv").read(), end="")
````

### Slide 53 — Read, group, average and write


````python
import csv

with open("grades.csv") as f:
    rows = list(csv.DictReader(f))

by_sec = {}
for r in rows:
    sec = r["section"]
    if sec not in by_sec:
        by_sec[sec] = []
    by_sec[sec].append(int(r["score"]))

with open("summary.csv", "w", newline="") as f:
    w = csv.writer(f)
    w.writerow(["section", "mean_score"])
    for sec in sorted(by_sec):
        w.writerow([sec, round(sum(by_sec[sec]) / len(by_sec[sec]), 1)])
````
