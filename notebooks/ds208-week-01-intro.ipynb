### Slide 21 — Arithmetic on whole arrays


Expected output:

```
[ 1.24  0.81  2.27 15.63 30.19 28.8 ]
131.6
[False False False  True  True  True]
[156.3 301.9 288. ]
```


````python
import numpy as np

# monthly rainfall, mm — one array, no loops
rain = np.array([12.4, 8.1, 22.7, 156.3, 301.9, 288.0])

print(rain * 0.1)            # convert all values to cm
print(round(rain.mean(), 1)) # the average
print(rain > 100)            # over 100 mm? one answer each
print(rain[rain > 100])      # just the wet months
````

### Slide 25 — A chart from a table in three lines


Setup for the next cell: the small made-up data its code reads.


````python
import io
import pandas as pd
df = pd.read_csv(io.StringIO("""month,visits
1,120
1,95
2,130
2,88
3,150
3,101
4,170
4,96"""))      # month 1 = January; two clinics per month
````

````python
import matplotlib.pyplot as plt

monthly = df.groupby("month")["visits"].sum()

monthly.plot(kind="bar")
plt.title("Clinic visits by month")
plt.show()   # the figure appears right under the code
````
