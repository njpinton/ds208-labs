## Part 0 &middot; A class bundles data

A **class** is a blueprint. `__init__` runs when you make one; `self` is *this
particular* object, and the attributes you set on it are its own. Run the cell.

```python
class Student:
    def __init__(self, sid, name):
        self.sid = sid
        self.name = name
        self.grades = []

ana = Student("2024-001", "Ana")
ben = Student("2024-002", "Ben")
ana.grades = [88, 92, 79]

print(ana.name, ana.grades)
print(ben.name, ben.grades)   # ben has his own, separate data
```

## Part 1 &middot; A method uses the object's own data

A **method** is a function that lives on the object and reaches its data through
`self`. Because the grades are already on `ana`, she can average them herself. Run the
cell.

```python
class Student:
    def __init__(self, sid, name):
        self.sid = sid
        self.name = name
        self.grades = []
    def average(self):
        return sum(self.grades) / len(self.grades)

ana = Student("2024-001", "Ana")
ana.grades = [88, 92, 79]
print("average:", round(ana.average(), 2))
```

**Answer here:**

1. `average()` takes no arguments beyond `self`, yet it knows Ana's grades. Where did the

## Part 2 &middot; Catch an error instead of crashing

Real input is messy. When a line might fail, wrap it in `try` and handle the exact
exception you expect in `except`. Run the cell.

```python
def to_score(raw):
    try:
        return int(raw)
    except ValueError:
        print(f"  '{raw}' is not a number - using 0")
        return 0

for raw in ["88", "ninety", "73", ""]:
    print(raw, "->", to_score(raw))
```

**Answer here:**

1. Which of the four inputs triggered the `except` branch, and what did `int()` do with
   them that made it raise `ValueError`?
   &rarr; *your answer*

2. A bare `except:` would also have caught everything &mdash; including typos in your own
   code. Why is naming `ValueError` specifically the safer choice?
   &rarr; *your answer*

## Part 3 &middot; Raise your own error on bad input

Don't let a nonsense value slip downstream. Check it at the door and `raise` a
`ValueError` with a message that says what went wrong. Run the cell.

```python
def set_grade(g):
    if not 0 <= g <= 100:
        raise ValueError(f"grade out of range: {g}")
    return g

print(set_grade(88))       # fine
print(set_grade(130))      # raises - read the last line
```

**Answer here:**

1. The second call stops the cell with a `ValueError`. How is *raising* an error here
   better than silently returning `130`?
   &rarr; *your answer*
